```python
import os
import shutil
import yaml
from concurrent.futures import ThreadPoolExecutor
from tqdm import tqdm
from ultralytics import YOLO

print("🛡️ INITIATING BULLETPROOF RESUME PROTOCOL (RELAY MODE)...")

# 1. Satisfy YOLO's memory
local_yaml_path = '/content/Dhaka_Traffic_Local.yaml'
with open(local_yaml_path, 'w') as f:
    yaml.dump({
        'path': '/content/Master_Dataset_Local',
        'train': 'images/train',
        'val': 'images/val',
        'test': 'images/test',
        'nc': 8,
        'names': {0: 'Bike', 1: 'Bus/Microbus', 2: 'Car', 3: 'CNG', 4: 'Cycle', 5: 'People', 6: 'Rickshaw', 7: 'Truck'}
    }, f, sort_keys=False)
print("✅ Configuration map locked and loaded.")

# 2. Smart Transfer Engine
drive_dir = '/content/drive/MyDrive/Thesis/Version 6/Master_Dataset'
local_dir = '/content/Master_Dataset_Local'

if not os.path.exists(os.path.join(local_dir, 'images', 'train')):
    print("📂 Local data missing (New Server). Initiating high-speed transfer from Drive...")
    files_to_copy = []
    for root, _, files in os.walk(drive_dir):
        for file in files:
            src_path = os.path.join(root, file)
            dst_path = src_path.replace(drive_dir, local_dir)
            files_to_copy.append((src_path, dst_path))

    for _, dst_path in files_to_copy:
        os.makedirs(os.path.dirname(dst_path), exist_ok=True)

    def copy_file(paths):
        src, dst = paths
        if not os.path.exists(dst):
            shutil.copy2(src, dst)

    with ThreadPoolExecutor(max_workers=32) as executor:
        list(tqdm(executor.map(copy_file, files_to_copy), total=len(files_to_copy), desc="Transferring Data"))
    print("✅ High-speed transfer complete.")
else:
    print("⚡ Local data already exists! Skipping transfer to save time.")

# 3. Automatic Resume
print("🚀 Waking up YOLO and Resuming Training from last checkpoint...")
model = YOLO('/content/drive/MyDrive/Thesis/Version 6/Training_Runs/Dhaka_YOLO26m_Baseline-4/weights/last.pt')
results = model.train(resume=True)
```

```python
from google.colab import drive
import torch
import os
from ultralytics import YOLO

print("🔌 Drive connection exhausted. Forcing a hard reset...")
drive.flush_and_unmount()
drive.mount('/content/drive', force_remount=True)

print("🚨 GOOGLE API BYPASS: Reading from Shared Drive, Saving to Personal Drive...")

original_ckpt_path = '/content/drive/MyDrive/Thesis/Version 6/Training_Runs/Dhaka_YOLO26m_Baseline-4/weights/last.pt'
patched_ckpt_path = '/content/last_patched.pt'
account_b_save_dir = '/content/drive/MyDrive/Relay_Training/Dhaka_YOLO26m_Baseline-4_Part2'

os.makedirs(account_b_save_dir, exist_ok=True)

print("🧠 Patching checkpoint...")
ckpt = torch.load(original_ckpt_path, map_location='cpu', weights_only=False)
ckpt['train_args']['save_dir'] = account_b_save_dir
ckpt['train_args']['project'] = '/content/drive/MyDrive/Relay_Training'
ckpt['train_args']['name'] = 'Dhaka_YOLO26m_Baseline-4_Part2'
torch.save(ckpt, patched_ckpt_path)
print("✅ Patch successful.")

print("🚀 Waking up YOLO and Resuming Training from Epoch 25...")
model = YOLO(patched_ckpt_path)
results = model.train(resume=True)
```
